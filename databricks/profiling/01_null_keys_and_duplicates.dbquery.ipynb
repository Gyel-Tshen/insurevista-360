```sql
SELECT 'policies' AS tbl, COUNT(*) AS total_rows, COUNT(DISTINCT policy_id) AS distinct_keys,
       SUM(CASE WHEN policy_id IS NULL THEN 1 ELSE 0 END) AS null_pk,
       SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END) AS null_fk
FROM hollard_bronze.policies
UNION ALL
SELECT 'claims', COUNT(*), COUNT(DISTINCT claim_id),
       SUM(CASE WHEN claim_id IS NULL THEN 1 ELSE 0 END),
       SUM(CASE WHEN policy_id IS NULL THEN 1 ELSE 0 END)
FROM hollard_bronze.claims
UNION ALL
SELECT 'customers', COUNT(*), COUNT(DISTINCT customer_id),
       SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END), NULL
FROM hollard_bronze.customers
UNION ALL
SELECT 'premium_transactions', COUNT(*), COUNT(DISTINCT transaction_id),
       SUM(CASE WHEN transaction_id IS NULL THEN 1 ELSE 0 END),
       SUM(CASE WHEN policy_id IS NULL THEN 1 ELSE 0 END)
FROM hollard_bronze.premium_transactions;
```
