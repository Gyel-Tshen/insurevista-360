```sql
SELECT 'claims → policies' AS check_name, COUNT(DISTINCT c.claim_id) AS orphans
FROM hollard_bronze.claims c LEFT ANTI JOIN hollard_bronze.policies p ON c.policy_id = p.policy_id
UNION ALL
SELECT 'premium → policies', COUNT(DISTINCT t.transaction_id)
FROM hollard_bronze.premium_transactions t LEFT ANTI JOIN hollard_bronze.policies p ON t.policy_id = p.policy_id
UNION ALL
SELECT 'policies → customers', COUNT(DISTINCT p.policy_id)
FROM hollard_bronze.policies p LEFT ANTI JOIN hollard_bronze.customers c ON p.customer_id = c.customer_id
UNION ALL
SELECT 'policies → brokers', COUNT(DISTINCT p.policy_id)
FROM hollard_bronze.policies p LEFT ANTI JOIN hollard_bronze.brokers b ON p.broker_id = b.broker_id;
```
