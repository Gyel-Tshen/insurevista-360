```sql
USE CATALOG dbw_insurevista_dev;

SELECT 'policies.policy_status' AS col, policy_status AS value, COUNT(*) AS n FROM hollard_bronze.policies GROUP BY ALL
UNION ALL
SELECT 'claims.claim_status', claim_status, COUNT(*) FROM hollard_bronze.claims GROUP BY ALL
UNION ALL
SELECT 'premium.payment_status', payment_status, COUNT(*) FROM hollard_bronze.premium_transactions GROUP BY ALL
UNION ALL
SELECT 'premium.payment_method', payment_method, COUNT(*) FROM hollard_bronze.premium_transactions GROUP BY ALL
ORDER BY col, value;
```
